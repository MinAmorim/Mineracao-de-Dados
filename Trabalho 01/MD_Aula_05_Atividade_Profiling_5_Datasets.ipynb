### Ficha técnica comparativa

Esta tabela é o **item 2 do mini-relatório**. Ela responde, de uma só vez, ao passo
"Estrutura" e a boa parte do passo "Qualidade" dos cinco conjuntos.

```python
def ficha_tecnica(df: pd.DataFrame) -> dict:
    """Resumo estrutural e de qualidade de um DataFrame."""
    n_linhas, n_colunas = df.shape
    numericas = df.select_dtypes(include="number").columns
    constantes = [c for c in df.columns if df[c].nunique(dropna=False) <= 1]
    celulas = n_linhas * n_colunas
    return {
        "linhas": n_linhas,
        "colunas": n_colunas,
        "numericas": len(numericas),
        "nao_numericas": n_colunas - len(numericas),
        "% ausentes": round(100 * df.isna().sum().sum() / celulas, 2),
        "duplicatas": int(df.duplicated().sum()),
        "col. constantes": len(constantes),
        "% zeros": round(100 * (df[numericas] == 0).sum().sum()
                         / max(n_linhas * len(numericas), 1), 2),
        "memoria (KB)": round(df.memory_usage(deep=True).sum() / 1024, 1),
    }


ficha = pd.DataFrame({nome: ficha_tecnica(df) for nome, df in dfs.items()}).T
ficha
```

## Parte 2 — Um perfil por dataset

A função abaixo gera e salva o relatório. Os argumentos seguem as recomendações
vistas em aula:

- `minimal=True` para conjuntos com muitas colunas (desliga correlações e
  interações caras);
- amostragem para conjuntos com muitas linhas.

> **Atenção:** `minimal=True` e amostragem aceleram o perfil, mas escondem
> justamente as correlações que interessam ao relatório. Use como primeira
> passada e **confirme no conjunto completo** antes de escrever os achados.

```python
def gerar_perfil(nome: str, df: pd.DataFrame, minimal: bool = False,
                 amostra: int | None = None, salvar: bool = True) -> ProfileReport:
    """Gera o ProfileReport de um dataset e salva em perfil_<nome>.html."""
    dados = df
    titulo = f"Perfil — {nome}"
    if amostra is not None and len(df) > amostra:
        dados = df.sample(amostra, random_state=42)
        titulo += f" (amostra de {amostra} linhas)"

    perfil = ProfileReport(dados, title=titulo, explorative=not minimal,
                           minimal=minimal)
    if salvar:
        perfil.to_file(f"perfil_{nome}.html")
        print(f"salvo: perfil_{nome}.html")
    return perfil


perfis = {}
```

## Parte 5 — Limites do profiling automático

O `ydata-profiling` calcula muita coisa, mas não sabe nada sobre o problema de
negócio. Aponte pelo menos **três** limitações que vocês encontraram na prática.
Alguns fios para puxar:

- A ferramenta consegue dizer se um valor extremo é **erro de medição** ou
  **fenômeno real**?
- Ela reconhece que uma coluna numérica é, na verdade, **categórica** ou um
  **identificador**?
- Ela detecta violação de **regra de negócio** (por exemplo, uma data de saída
  anterior à de entrada)?
- Correlação alta entre um atributo e o alvo é sinal de bom preditor ou de
  **vazamento**? Quem decide?

**Limitação 1: A ferramenta não distingue erro de medição de fenômeno real**

No California Housing, o profiling da amostra de 5.000 linhas classificou AveOccup como assimétrica, mas não consegue dizer se o valor máximo (83 na amostra e 1243 no dataset completo) é um erro ou um caso real. A ferramenta apenas identifica uma distribuição incomum. Cabe ao analista entender a causa e decidir como tratar o problema.

**Limitação 2: A ferramenta não reconhece sozinha quando um número representa uma categoria**

No Diabetes, sex é uma variável numérica, com valores como -0,0446 e 0,0507 devido à normalização. O ydata-profiling identificou a variável como categórica porque ela possui apenas dois valores distintos. Ou seja, a classificação foi baseada na quantidade de valores, e não no significado da variável. Essa é uma limitação importante, pois a ferramenta pode classificar uma variável de forma incorreta quando a cardinalidade não representa bem sua função.

**Limitação 3: Alta correlação pode indicar um bom preditor ou vazamento de dados**

No Iris e no Wine, encontramos correlações altas entre variáveis e o alvo (petal width x target = 0,924; flavanoids x target = 0,752), que fazem sentido e podem indicar bons preditores. Porém, a ferramenta não consegue diferenciar uma correlação legítima de um possível vazamento de dados. Se uma variável fosse criada depois do resultado ser conhecido, por exemplo, o profiling também destacaria a alta correlação sem identificar o vazamento. Essa decisão depende do analista e do conhecimento sobre o domínio.

**Limitação 4: Métricas agregadas podem esconder o significado real da coluna**

No Iris, a ficha técnica mostrou 6,67% de zeros. Isso não representa dados faltantes ou valores inválidos, pois o target usa o valor 0 para representar a classe setosa. A métrica está correta, mas sem conhecer o significado da coluna pode parecer um problema de qualidade. Isso mostra que as ferramentas automáticas descrevem os dados, mas não entendem completamente seu significado.
